```python
def load_data():
    df = pd.read_csv('src/data/processed_moviesdrama.csv')
    df["release_year"] = pd.to_numeric(df["release_year"], errors="coerce")
    df["score"] = pd.to_numeric(df["score"], errors="coerce")
    df["id"] = df["id"].astype(str)  # keep id as string for consistency
    df["primary_genre"] = (df['genre'].str.split(",").str[0].str.strip())

    # new dataframe for searching purpose
    drama = df[df["type"] == "Drama"].copy()
    movie = df[df["type"] == "Movie"].copy()
    return df, drama, movie
```

```python
df, _, _ = load_data()
```

```python
## Spotlight

new_release = df[df['release_year' >= 2026]]
todays_pick = df[(df["score"]>7) & (df["release_year"] >= 2025)]
top_rated = df[(df["score"] >= 8.5) & (df["user_count"] > 1000)]
most_popular = df[df["user_count"] >= 30000]
```
